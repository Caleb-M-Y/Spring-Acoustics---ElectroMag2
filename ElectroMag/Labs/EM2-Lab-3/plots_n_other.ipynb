```python
import matplotlib.pyplot as plt
import numpy as np

# Data setup
categories = ['Coax (Open)', 'Coax (Short)', 'Twin-Lead (Open)', 'Twin-Lead (Short)']
v_inc = [-3.56, 460.0, 145.6, 698.9]
v_refl = [-181.6, -345.0, -247.0, -276.6]

x = np.arange(len(categories))  # the label locations
width = 0.35  # the width of the bars

# Create the plot
fig, ax = plt.subplots(figsize=(9, 6))
rects1 = ax.bar(x - width/2, v_inc, width, label='Incident Voltage ($V_{inc}$)', color='#1f77b4', edgecolor='black')
rects2 = ax.bar(x + width/2, v_refl, width, label='Reflected Voltage ($V_{refl}$)', color='#ff7f0e', edgecolor='black')

# Add labels, title, and formatting
ax.set_ylabel('Voltage (mV)', fontsize=12, fontweight='bold')
ax.set_title('Incident vs. Reflected Voltage Amplitudes across Terminations', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11)
ax.axhline(0, color='black', linewidth=1) # Adds a line at Y=0 for visual clarity
ax.legend(fontsize=11)

# Add gridlines for readability
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Make the layout tight and save the figure
plt.tight_layout()
plt.savefig('voltage_reflection_chart.png', dpi=300)

# Show the plot
plt.show()
```
