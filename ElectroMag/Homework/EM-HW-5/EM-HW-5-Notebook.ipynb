## Problem 5: The Rainbow
Here we work through the multiple reflections and refractions that give rise to the rainbow. Consider a ray of sunlight incident from the left a distance $h$ above the $x$-axis on a spherical drop of water with radius 1 centered at the origin such that $0 < h < 1$, as illustrated in Figure 1. The index of refraction for air, $n_{\text{air}} = 1$, and use your index of refraction values for red and blue light from a previous problem.



**Figure 1 NOTE:** the indices of refraction are $n_{\text{red}} = 1.3$ for red light (700nm) and $n_{\text{blue}} = 1.4$ for blue light (400nm) in this figure for illustration purposes only.

-------------------------------------------------------------------------------------
**(a) Point of Entry and Surface Normal:** The entry point on the circle is,

$$\boldsymbol{r}_1 = -(\sqrt{R^2 - h^2})\hat{x} + h\hat{y}$$

The outward unit normal at this point is:

$$\hat{n} = \frac{\boldsymbol{r}_1}{R} = -\frac{\sqrt{R^2 - h^2}}{R}\hat{x} + \frac{h}{R}\hat{y}$$

Pick a value of $h$ in the range $0 < h < 1$ and calculate the vectors $\boldsymbol{r}_1$ and $\hat{n}$ for $R = 1$.

-------------------------------------------------------------------------------------
**(b) Incident Angle & Transmitted Angle:** The incoming sunray is in the direction,

$$\boldsymbol{v}_{\text{in}} = +1\hat{x} + 0\hat{y}$$

Determine the angle of incidence, $\theta_i$, by computing the angle between the incoming ray direction (from the Sun) and the outward normal using the dot product formula:

$$\theta_i = \arccos(-\boldsymbol{v}_{\text{in}} \cdot \hat{n})$$

(Note: we use the negative dot product since the ray comes from outside the drop.) Calculate $\theta_i$ for your choice of $h$.



Use Snell's Law to calculate the transmission angle:

$$n_{\text{air}} \sin\theta_i = n_{\text{red}} \sin\theta_t \quad \Rightarrow \quad \theta_t = \arcsin\!\left(\frac{n_{\text{air}}}{n_{\text{red}}}\sin\theta_i\right)$$

Compute $\theta_t$ for red light for your value of $\theta_i$.

-------------------------------------------------------------------------------------
**(c) Direction of Refracted Ray:** Rotate the inward-pointing surface normal by the transmission angle to construct the new direction vector for the transmitted ray inside the drop. The inward normal is $-\hat{n}$. We rotate this vector **counter-clockwise** by $\theta_t$ to obtain the transmitted ray direction using matrix multiplication:

$$\boldsymbol{v}_{\text{trans}} = \mathcal{R}(\theta_t)(-\hat{n})$$

with the 2D rotation matrix:

$$\mathcal{R}(\theta) = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}$$

and

$$-\hat{n} = \frac{\sqrt{R^2 - h^2}}{R}\hat{x} - \frac{h}{R}\hat{y} = \begin{bmatrix} \sqrt{R^2 - h^2}/R \\ -h/R \end{bmatrix}$$

Compute $\boldsymbol{v}_{\text{trans}}$ using your calculated value of $\theta_t$.

-------------------------------------------------------------------------------------
**(d) First Internal Reflection Location:** Recall $\boldsymbol{r}_1$ is the point where the ray enters the raindrop. Let $\hat{t}$ be the unit vector pointing in the direction of the ray after it has refracted inside the drop, in the direction of $\boldsymbol{v}_{\text{trans}}$:

$$\hat{t} = \frac{\boldsymbol{v}_{\text{trans}}}{\|\boldsymbol{v}_{\text{trans}}\|}$$

Compute $\hat{t}$ from $\boldsymbol{v}_{\text{trans}}$. We now wish to find the point $\boldsymbol{r}_2$ where this ray next intersects the boundary of the raindrop. Since the drop is a circle of radius $R = 1$ centered at the origin, the ray inside follows a parametric path of the form:

$$\boldsymbol{x}(s) = \boldsymbol{r}_1 + s\hat{t}, \quad s > 0$$

The intersection point is on the circle so $\|\boldsymbol{x}(s)\|^2 = R^2 (= 1)$. Substituting and expanding gives a quadratic equation in $s$:

$$s^2 + 2(\boldsymbol{r}_1 \cdot \hat{t})s + (\|\boldsymbol{r}_1\|^2 - R^2) = 0$$

Solving this gives two values of $s$:

$$s = \frac{-2(\boldsymbol{r}_1 \cdot \hat{t}) \pm \sqrt{[2(\boldsymbol{r}_1 \cdot \hat{t})]^2 - 4(\|\boldsymbol{r}_1\|^2 - R^2)}}{2}$$

with $R = 1$. The **smaller root**, $s_1$, corresponds to the original point $\boldsymbol{r}_1$. The **larger root**, $s_2$, gives the location of the second intersection:

$$\boldsymbol{r}_2 = \boldsymbol{r}_1 + s_2\hat{t}$$

Compute the roots $s_1$ and $s_2$. Use $s_2$ to determine $\boldsymbol{r}_2$.

-------------------------------------------------------------------------------------
**(e) Internal Reflection:** At this new point $\boldsymbol{r}_2$, the **interior** surface normal is:

$$\hat{n}_2 = -\frac{\boldsymbol{r}_2}{\|\boldsymbol{r}_2\|} = -\frac{\boldsymbol{r}_2}{R}, \quad \text{with } R = 1$$

Compute $\hat{n}_2$ at $\boldsymbol{r}_2$. We then apply the vector reflection formula (previous problem) to compute the new direction:

$$\hat{r} = \hat{t} - 2(\hat{t} \cdot \hat{n}_2)\hat{n}_2$$



This reflected direction $\hat{r}$ is the direction in which the ray will travel, $\boldsymbol{v}_{\text{refl}}$, toward the opposite wall of the drop for potential refraction and exit.

-------------------------------------------------------------------------------------
**(f) Exit Refraction:** Let $\boldsymbol{r}_3$ be the third intersection point on the circle. Find this point via a similar method as in part (d). At this point, compute the new angle of incidence $\theta_{i,3}$ relative to the inward surface normal $-\hat{n}_3 = -\boldsymbol{r}_3/R$, with $R = 1$:

$$\theta_{i,3} = \arccos\!\left(-\boldsymbol{v}_{\text{refl}} \cdot (-\hat{n}_3)\right)$$

where $\boldsymbol{v}_{\text{refl}} = \boldsymbol{v}_{\text{in}}$ in Figure 4.



Then use Snell's Law to compute the transmission angle:

$$\theta_{t,3} = \arcsin\!\left(\frac{n_{\text{red}}}{n_{\text{air}}}\sin\theta_{i,3}\right)$$

The final exit direction is computed by rotating the normal vector $\hat{n}_3$ **clockwise** by $\theta_{t,3}$ via matrix multiplication:

$$\boldsymbol{v}_{\text{exit}} = \mathcal{R}(-\theta_{t,3})\hat{n}_3$$

Note: Clockwise rotations are negative angles. Counter-clockwise rotations are positive angles. Determine $\theta_{t,3}$ and $\boldsymbol{v}_{\text{exit}}$.

-------------------------------------------------------------------------------------
**(g) Scattering Angle (Rainbow Angle):** The observable scattering angle between the incident sunlight direction and the exiting ray is called the **deviation angle**, denoted $\Theta_D$. We define the deviation angle as the angle between the reversed incident direction ($-\boldsymbol{v}_{\text{incident}}$) and the final exit direction $\boldsymbol{v}_{\text{exit}}$:

$$\Theta_D = \cos^{-1}\!\left(\frac{(-\boldsymbol{v}_{\text{incident}}) \cdot \boldsymbol{v}_{\text{exit}}}{\|\boldsymbol{v}_{\text{incident}}\| \cdot \|\boldsymbol{v}_{\text{exit}}\|}\right)$$

Since both vectors are unit length in our construction, this simplifies to:

$$\Theta_D = \cos^{-1}(-\boldsymbol{v}_{\text{incident}} \cdot \boldsymbol{v}_{\text{exit}})$$

Plot $\Theta(h)$, and determine the extremum of the deviation function for red light. Repeat for blue light. Which is deviated more?
